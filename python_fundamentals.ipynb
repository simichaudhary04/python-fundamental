3.  What are operators in Python? Explain different types with
examples.
   - Operator: Operators are symbols used to perform operations on variables and values.
   
   Types of Operators:
   
   1. Arithmetic Operator: Used for maths.
       +, -, _, /, %, __
     Ex: a=10, b=3 -> a+b=13, a%b=1.
     
    2.  Assignment Operator: Used to assign value.
        =, +=, -=, _=
     Ex: x=5, x+=2 -> x=7
     
     3. Comparison Operator: Used to compare, returns True/False.
        ==, !=, >, <, >=, <=
     Ex: 5==5 -> True, 5>3 -> True
     
     4. Logical Operator: Used for logic.
        and, or, not
     Ex: True and False -> False5. Membership Operator: Checks in sequence.
       in, not in
      Ex: "a" in "Devops" -> True6. Identity Operator: Checks same object.
         is, is not
       Ex: a is b

       

4.  How does the if-else statement work in Python? Explain with a
practical DevOps example.
  - if-else: It is used for decision making. If condition is True, if block runs, else else block runs.
  
   Syntax:
   
   if condition:
     # runs if True
   else:
     # runs if False

     example:

     disk_usage = 85

     if disk_usage > 80:
        print("Warning: Disk full, cleanup needed")
     else:
        print("Disk is OK")


5.  Explain loops in Python and describe how they can be used in
DevOps automation.
   - Loop: A loop is used to repeat a block of code multiple times.


   1. for loop - Iterates over a collection or range (list, tuple etc.)
   
   for i in range(5):
       print(i)
    
    2. while loop - Repeats a block of code as long as a condition is True.
    
       x=1
     while x < 5:
       print(x)
      x+=1
      
      Loop Control Statements:
      Break - Exits the loop completely.
      
      Continue - Skips the current iteration and moves to the next.
      
      pass - Is a placeholder for code that does nothing.
      
      Use in DevOps Automation:
   In DevOps, loops are used to automate repetitive tasks on multiple servers.
   
   servers = ["server1", "server2", "server3"]
     for server in servers:
          print(f"Deploying app to {server}")
    
     This helps in checking health, deploying code, and cleaning logs on many servers at once.
     

6.  What are functions and modules in Python? Why are they
important in DevOps?
    - Function: A function is a block of reusable code that does a specific task. It helps to avoid repeating code.
    
      def check_server():
          print("Checking server health...")

     check_server() # function call

   Module: A module is a file containing Python functions and code. We use import to use it.

      import os
     import shutil
     print(os.name)
     
     Why important in DevOps?
     
     Code Reuse: Write once, use many times. Like one function to check all servers.
     
     Automation: We can use modules like os, subprocess, boto3 to automate server, docker, and AWS tasks.
     
     Clean Code: Big DevOps scripts become easy to manage when divided into functions and modules
